# use 1D, 2D and 3D np-array, and apply indexing and slicing atleast 5 example for each (take example as your wish)

```python
import numpy as np

# --- Array Creation ---
arr1d = np.arange(10)                  # 1D array: 10 elements
arr2d = np.arange(12).reshape(3, 4)    # 2D array: 3x4 matrix
arr3d = np.arange(24).reshape(2, 3, 4) # 3D array: 2x3x4 tensor

print("--- 1D Array Indexing & Slicing ---")
print(f"Original 1D: {arr1d}")
print(f"1. Single element [3]: {arr1d[3]}")
print(f"2. Negative index [-1]: {arr1d[-1]}")
print(f"3. Basic slice [2:7]: {arr1d[2:7]}")

print("--- 2D Array Indexing & Slicing ---")
print(f"Original 2D:\n{arr2d}")
print(f"1. Single element [1, 2]: {arr2d[1, 2]}")
print(f"2. Entire row [0, :]: {arr2d[0, :]}")
print(f"3. Entire column [:, 1]: {arr2d[:, 1]}")

print("--- 3D Array Indexing & Slicing ---")
print(f"Original 3D:\n{arr3d}")
print(f"1. Single element [0, 1, 2]: {arr3d[0, 1, 2]}")
print(f"2. Full 2D slice [1, :, :]:\n{arr3d[1, :, :]}")
print(f"3. 2D slice fixing middle dim [:, 1, :]:\n{arr3d[:, 1, :]}")
```

# use hstack() and vstack()

```python
arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([4, 5, 6, 8])

# Horizontal stack (joins along columns / axis 1)
h_stacked = np.hstack((arr1, arr2))

# Vertical stack (joins along rows / axis 0)
v_stacked = np.vstack((arr1, arr2))

print(f"hstack (Array 1 + Array 2):\n{h_stacked}\n")
print(f"vstack (Array 1 over Array 2):\n{v_stacked}\n")
```

# use hsplit() and vsplit()

```python
# Horizontal split (splits along columns into 2 equal parts)
h_split = np.hsplit(h_stacked, 2)

# Vertical split (splits along rows into 2 equal parts)
v_split = np.vsplit(v_stacked, 2)

print(f"h_split:\n{h_split}\n")
print(f"v_split:\n{v_split}\n")
```
