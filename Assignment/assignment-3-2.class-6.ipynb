# Merging and joining

```python
students = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115, 116, 117, 118, 119, 120],
    "name": ["Rahim", "Karim", "Hasan", "Sakib", "Nabil",
             "Tanvir", "Fahim", "Rafi", "Imran", "Shakil",
             "Arif", "Jahid", "Mehedi", "Siam", "Tareq",
             "Rakib", "Adnan", "Nayeem", "Foysal", "Rony"],
    "department": ["CSE", "EEE", "CSE", "BBA", "CSE",
                   "EEE", "CSE", "BBA", "CSE", "EEE",
                   "CSE", "BBA", "EEE", "CSE", "BBA",
                   "CSE", "EEE", "CSE", "BBA", "EEE"]
})

students
```

```python
results = pd.DataFrame({
    "student_id": [101, 102, 103, 105, 106, 108, 109, 111, 112, 114,
                   121, 122, 123, 124, 125, 126, 127, 128, 129, 130],
    "math": [85, 72, 91, 68, 77, 88, 74, 95, 81, 69,
             78, 84, 92, 73, 89, 67, 80, 94, 76, 83],
    "python": [90, 75, 88, 72, 80, 92, 79, 96, 85, 71,
               82, 87, 95, 76, 91, 70, 84, 93, 78, 86]
})

results
```

```python
departments = pd.DataFrame({
    "dept_id": [1, 2, 3, 4],
    "department": [
        "CSE",
        "EEE",
        "BBA",
        "Civil"
    ],
    "faculty": [
        "Engineering",
        "Engineering",
        "Business",
        "Engineering"
    ]
})

departments
```

# 1. Create a dataframe contains students name,faculty and number on math and python

```python
merged_table = pd.merge(
    pd.merge(
        students,
        departments,
        left_on='department',
        right_on='department',
        how='inner'
    ),
    results,
    left_on='student_id',
    right_on='student_id',
    how='left'
)

merged_table[['name', 'faculty', 'math', 'python']]
```

# 9. A new column contains previous 10 days average meantemp

```python
import numpy as np

step_len = 10

list_len = len(climate_time_series)
data_table = climate_time_series.copy()

# Initialize the new column
data_table['avg_10_days_meanpressure'] = 0.0

# Use range(start, stop, step) to automatically jump by 10s
for index in range(0, list_len, step_len):
    # Calculate the mean (now a single float value)
    mean_val = climate_time_series['meanpressure'].iloc[index : index + step_len].mean()
    
    # Assign that single float to all rows in the 10-row slice
    data_table.iloc[index : index + step_len, data_table.columns.get_loc('avg_10_days_meanpressure')] = mean_val

# View the result
data_table.head(20)
```
