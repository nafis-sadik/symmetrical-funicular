### Exercise 1.1: Temperature Converter (Celsius to Fahrenheit)
Write code to convert temperature from Celsius to Fahrenheit using the formula:
$$\text{Fahrenheit} = (\text{Celsius} \times 9/5) + 32$$

- Define a variable `celsius = 25`
- Calculate `fahrenheit`

### Exercise 11: `while` Loop Counter
Use a `while` loop to print numbers from 1 up to 5.
Inside the loop, when the counter reaches 3, also print `"Reached number 3!"`.


````python
count = 1

while count <= 5:
    print("Count:", count)
    if count == 3:
        print(" -> Reached number 3!")
    count += 1
````

### Exercise 12: Loop Control (`break` and `continue`)
Loop through numbers from 1 to 10 using `for num in range(1, 11)`:
1. If `num == 4`, skip printing it using `continue`.
2. If `num == 8`, stop the loop entirely using `break`.
3. Print each number that is not skipped.


````python
for num in range(1, 11):
    if num == 4:
        continue  # Skip 4
    if num == 8:
        print("Breaking loop at:", num)
        break     # Stop loop
    print("Number:", num)
````

### Exercise 18: Sets (Removing Duplicates & Operations)
1. Given a list with duplicates: `numbers = [1, 2, 2, 3, 4, 4, 5, 5, 5]`.
   - Convert it to a set named `unique_numbers` and print it.
2. Given two sets:
   `set_a = {1, 2, 3, 4}`
   `set_b = {3, 4, 5, 6}`
   - Find and print their **Union** (`set_a | set_b`).
   - Find and print their **Intersection** (`set_a & set_b`).


````python
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 5]
unique_numbers = set(numbers)
print("Unique Numbers Set:", unique_numbers)

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print("Union (A | B)       :", set_a | set_b)
print("Intersection (A & B):", set_a & set_b)
````

### Exercise 19: Dictionary Operations
Given the dictionary:
`product = {"name": "Laptop", "price": 1200, "brand": "Dell"}`

1. Access and print the `"name"` and `"price"`.
2. Add a new key `"in_stock"` with value `True`.
3. Update the `"price"` to `1150`.
4. Use a `for` loop to iterate through `product.items()` and print every key and value.


````python
product = {"name": "Laptop", "price": 1200, "brand": "Dell"}

# 1. Access
print("Product Name :", product["name"])
print("Product Price:", product["price"])

# 2. Add new key
product["in_stock"] = True

# 3. Update price
product["price"] = 1150

# 4. Iterate and print
print("--- Product Details ---")
for key, value in product.items():
    print(f"{key}: {value}")
````

### Exercise 20.1: Squaring Numbers
Given `numbers = [1, 2, 3, 4, 5]`:

### Exercise 20.2: Filtering Even Numbers
Given `nums = [12, 7, 18, 23, 30, 41, 50]`:
- Use a **list comprehension with an `if` condition** to filter and keep only the **even numbers** (`x % 2 == 0`).
- Print the filtered list (Expected: `[12, 18, 30, 50]`).


````python
nums = [12, 7, 18, 23, 30, 41, 50]
even_nums = [x for x in nums if x % 2 == 0]
print("Even Numbers:", even_nums)
````
